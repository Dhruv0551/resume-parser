```python
from groq import Groq
from dotenv import load_dotenv
from pydantic import BaseModel
from pypdf import PdfReader
import os
import json
```

```python
load_dotenv()
api_key = os.getenv("GROQ_API_KEY")
if not api_key:
    raise ValueError("No Api Key Found")

client = Groq(api_key=api_key)
model = "llama-3.3-70b-versatile"
```

Defining Classes & Schemas

```python
class Job_D(BaseModel):
    role: str
    required_skills: list[str]
    prefered_skills: list[str]
    minimum_experience: float or None
    educational_requirement: list[str]
    responsibility: list[str]

Job_D_Schema = Job_D.model_json_schema()

class Expeirence(BaseModel):
    role: str
    duration: float
    used_skills: list[str]

Expeirence_Schema = Expeirence.model_json_schema()

class Resume(BaseModel):
    name: str
    education: list[str]
    skills: list[str]
    projects: list[str] or None
    expeirence: list[Expeirence] or None
    Achievements: list[str] or None

Resume_Schema = Resume.model_json_schema()
```
